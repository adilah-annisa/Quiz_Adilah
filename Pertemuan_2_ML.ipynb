```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Selamat! Semua berhasil diimpor")
```

```python
data = {
    "Nama": ["Dila", "Ayu", "Beni", "Doel", "Jack"],
    "Program_Studi": ["TI","SI","TI","SI","TRK"],
    "Waktu_Belajar":[4,6,3,5,2],
    "Kehadiran":[90,75,80,88,70],
    "Nilai":[90,75,70,80,82]
}

df = pd.DataFrame(data)
df
```

```python
df.describe()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    df["Nama"],
    df["Nilai"],
    color="red"
)

plt.title("Nilai Ujian Mahasiswa")
plt.xlabel("Nama Mahasiswa")
plt.ylabel("Nilai Ujian")

plt.ylim(0, 100)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

```python
epoch = [1, 2, 3, 4, 5]
akurasi = [60, 68, 75, 82, 88]
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    epoch,
    akurasi,
    marker="o",
    color="red",
    linestyle="--"
)

plt.title("Perkembangan Akurasi Model")
plt.xlabel("Epoch")
plt.ylabel("Akurasii (%)")

plt.xticks(epoch)
plt.ylim(0, 100)
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Nilai"],
    bins=5,
    color="skyblue",
    edgecolor="black"
)

plt.title("Distribusi Nilai Ujian Mahasiswa")
plt.xlabel("Nilai Ujian")
plt.ylabel("Frekuensi")

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Waktu_Belajar"],
    df["Nilai"],
    color="purple",
    s=85
)

plt.title("Hubungan Waktu Belajar dan Nilai Ujian")
plt.xlabel("Waktu Belajar ( jam)")
plt.ylabel("Nilai Ujian")

plt.xlim(0, 9)
plt.ylim(0, 100)
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Waktu_Belajar",
    y="Nilai",
    hue="Program_Studi",
    s=105
)

plt.title("Waktu Belajar dan Nilai Berdasarkan Program Studi")
plt.xlabel("Waktu Belajar ( jam)")
plt.ylabel("Nilai Ujian")

plt.show()
```

```python
# Memilih kolom numerik
data_numerik = df[
    ["Waktu_Belajar", "Kehadiran", "Nilai"]
]

# Menghitung korelasi
korelasi = data_numerik.corr()

# Membuat Heatmap
sns.heatmap(
    korelasi,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Heatmap Korelasi Variabel Mahasiswa")
plt.show()
```

```python
# LATIHAN MACHINE LEARNING
# 1 Upload File CSV
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Membaca dataset
file_path = "/content/ecommerce_sales_analytics_5000.csv"

if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    print("Dataset berhasil dibaca")
    print(df.columns.tolist())
else:
    print("File tidak ditemukan.")
    print("File yang ada di Colab:")
    print(os.listdir("/content"))
```

```python
#2. Grafik Batang
revenue_kategori = df.groupby("product_category")["revenue"].sum().reset_index()

plt.figure(figsize=(8, 5))

plt.bar(
    revenue_kategori["product_category"],
    revenue_kategori["revenue"],
    color="green"
)

plt.title("Total Revenue Berdasarkan Kategori Produk")
plt.xlabel("Kategori Produk")
plt.ylabel("Total Revenue")

plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
#3. Grafik Donat
# Menghitung total revenue berdasarkan metode pembayaran
revenue_payment = df.groupby("payment_method")["revenue"].sum()

plt.figure(figsize=(7, 7))

plt.pie(
    revenue_payment,
    labels=revenue_payment.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.4}
)

plt.title("Persentase Revenue Berdasarkan Metode Pembayaran")

plt.show()
```

```python
# 4. Grafik Garis
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df["order_date"] = pd.to_datetime(df["order_date"])

# Total revenue per bulan
revenue_bulanan = (
    df.set_index("order_date")
      .resample("MS")["revenue"]
      .sum()
      .reset_index()
)

plt.figure(figsize=(12, 5))

plt.plot(
    revenue_bulanan["order_date"],
    revenue_bulanan["revenue"],
    color="red",
    linestyle="--"
)

plt.title("Perkembangan Revenue Berdasarkan Bulan")
plt.xlabel("Bulan")
plt.ylabel("Total Revenue")

# Tampilkan label hanya setiap 6 bulan
plt.gca().xaxis.set_major_locator(mdates.MonthLocator(interval=6))
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))

plt.xticks(rotation=45)

plt.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
# 5. Scatter Plot
df_scatter = df.sample(
    n=min(300, len(df)),
    random_state=42
)

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_scatter,
    x="unit_price",
    y="revenue",
    hue="product_category",
    s=25,
    alpha=0.7
)

plt.title("Hubungan Harga Produk dan Revenue Berdasarkan Kategori")
plt.xlabel("Harga Satuan Produk")
plt.ylabel("Revenue")

plt.legend(
    title="Kategori",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.grid(True, linestyle="--", alpha=0.3)
plt.tight_layout()

plt.show()
```
